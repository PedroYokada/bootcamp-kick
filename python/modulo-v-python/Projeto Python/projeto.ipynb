```python
## GRÁFICO TURNOS GRADUAÇÃO


# Um dicionario data é criado para armazenar strings do campo 'TURNO' com os tipos de turnos, 
# alem disso o campo 'VALORES' para armazenar a quantidade de estudantes inscritos no tipo de turno do curso que escolheram.
data = {
    'TURNO': ['Integral', 'Noturno', 'Matutino', 'Vespertino', 'EAD'], 
    'Valores': [121699, 39215, 12315, 9539, 1439]
} 

# Criação do DataFrame turno, com os dados do dicionario (data).
turno = pd.DataFrame(data)

# Ocorre a criação de uma figura, alem disso (8,6) são as polegadas do gráfico
plt.figure(figsize=(8, 6)) 

# Um gráfico é criado com o dataframe data, de cor azul marinho
bars = plt.bar(turno['TURNO'], turno['Valores'], color='navy')

# Este laço de repetição passa por cada barra individualmente
for bar in bars:
    # plt.text() é uma função para adicionar o texto ao gráfico.
    # bar.get_x() define a posição x da barra.
    # bar.get_width()/2 pega a largura da barra e divide por 2 para centralizar o texto na barra
    # inclui separadores de milhar nos valores do gráfico
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height(), f'{bar.get_height():,}', 
             ha='center', va='bottom', fontsize=9)

# esta linha define a fonte, cor e outros atributos desta parte do PDF
font = {'family': 'arial', 'color': 'black', 'weight': 'normal', 'size': 9}
# Label do eixo X do gráfico.
plt.xlabel('TURNOS DOS CURSOS DE GRADUAÇÃO', fontdict=font)
# Label do eixo Y do gráfico.
plt.ylabel('NUMERO DE ALUNOS', fontdict=font)
# Label do titulo do gráfico.
plt.title('Distribuição de Turnos dos Cursos de Graduação', fontdict={'size': 14, 'weight': 'bold'}) 

# Criação da imagem turno.
plt.savefig('turno.png', bbox_inches='tight')
# fecha a imagem turno.
plt.close()
# Adiciona uma nova página ao pdf.
yok.add_page()
# Define as dimensões da imagem.
yok.image('turno.png', x=15, y=15, w=180)
```

```python
# Página TIPO DE GRADUAÇÃO


# Criação do dicionario grau_data, com as strings do tipo de grau do curso, alem com valores de cada grau.
grau_data = {'GRAU': ['Bacharelado', 'Licenciatura', 'Tecnológico', 'ABI'], 
             'GRADUACAO': [149171, 23285, 9368, 2383]} 

# criação do dataframe grau com os valores do dicionario grau_data
grau = pd.DataFrame(grau_data)

# Essa função serve para formatar um valor para percentual.
def format_pct(pct, allvalues):
    return "{:.1f}%".format(pct)



# Paleta de cores do gráfico em pizza 
# PALETA DE CORES DE CEU AZUL
cores = ['#60D4EA', '#9AE5F3', '#BBF3F9', '#A8DEE2']  

# Ocorre a criação de uma figura, alem disso (8,8) são as polegadas do gráfico
plt.figure(figsize=(8, 8))  

# Ocorre a criação gráfico em pizza, usando os dados do metodo grau
# Com os dados do dicionario grau_data
# A função format_pct é utilizado para identar os valores da GRADUACAO
# colors é usado para pegar as cores do metodo 'cores'
grau.set_index('GRAU')['GRADUACAO'].plot(kind='pie', 
autopct=lambda pct: format_pct(pct, grau['GRADUACAO']), 
colors=cores, 
legend=False,
startangle=90)  

# Ocorre a criação do titulo do gráfico.
plt.title('Distribuição dos Tipos de Cursos', fontdict=font, fontsize=14)

# Label Y vazia
plt.ylabel('') 

# Criação da imagem grau
plt.savefig('grau.png', bbox_inches='tight') 

# Imagem sendo fechada
plt.close()

# Criação de uma nova pagina
yok.add_page()
# Definição de fonte e tamanho.
yok.set_font("Times", size=12)
# Redimensionamento da imagem grau.
yok.image('grau.png', x=15, y=40, w=180) 


# Criação do Metodo multi_cell.
yok.multi_cell(
    0,  
    10,  
    "       O gráfico abaixo é um complemento do raciocínio anterior, de que a grande maioria dos alunos optaram por cursar cursos de bacharelado, em sua grande maioria o turno é integral.",
    align='J' 
)
```
